## Navios simultâneos

```python
# Define the optimized function
def count_waiting_ships(group):
    arrival_times = group['Data Chegada'].values
    berthing_times = group['Data Atracação'].values
    waiting_counts = []

    for i in range(len(arrival_times)):
        # Count ships that arrived before or at the same time and are still waiting
        waiting_count = ((arrival_times <= arrival_times[i]) & (berthing_times > arrival_times[i])).sum() - 1
        waiting_counts.append(max(0, waiting_count))

    group['Navios em Fila na Chegada'] = waiting_counts
    return group
```

```python
# Sort the DataFrame for consistent processing
df_merged_top5_simult = df_merged_top5.sort_values(by=['IDBerco', 'Data Chegada'])

# Apply the function to each group of 'IDBerco'
df_merged_top5_simult = df_merged_top5_simult.groupby('IDBerco', group_keys=False).apply(count_waiting_ships)

# Display the result
print(df_merged_top5_simult[['IDBerco', 'Data Chegada', 'Data Atracação', 'Navios em Fila na Chegada']].head())
```

```python
import pandas as pd
import numpy as np

#  Features e target
X = df_merged_simult_top5[['VLPesoCargaBruta', 'Natureza da Carga', 'Complexo Portuário', 'Tipo Operação da Carga', 'Sentido', 'Navios em Fila na Chegada']]
y = df_merged_simult_top5['Faixa_Tempo_Espera']

#  Dividir em treino e teste
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=9, stratify=y
)


# Combina X_train e y_train para facilitar
train_data = X_train.copy()
train_data['Faixa_Tempo_Espera'] = y_train

# Ver quantas amostras tem por classe
class_counts = train_data['Faixa_Tempo_Espera'].value_counts()
print("Distribuição original:")
print(class_counts)

# Define a quantidade de amostras por classe
# Pode ser a menor classe ou outro valor
min_samples = class_counts.min()

# Lista para armazenar os DataFrames reduzidos
dfs = []

# Para cada classe, sorteia 'min_samples' aleatórios
for cls in class_counts.index:
    subset = train_data[train_data['Faixa_Tempo_Espera'] == cls]
    sampled_subset = subset.sample(n=min_samples, random_state=9)
    dfs.append(sampled_subset)

# Concatena tudo
balanced_train_data = pd.concat(dfs)

# Separa de volta X e y
X_resampled = balanced_train_data.drop('Faixa_Tempo_Espera', axis=1)
y_resampled = balanced_train_data['Faixa_Tempo_Espera']

print("\nDistribuição após undersampling:")
print(y_resampled.value_counts())
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt


#  Treinamento com dados balanceados
logr = LogisticRegression(max_iter=2000)

logr.fit(X_resampled, y_resampled)

#  Predição
y_pred = logr.predict(X_test)

#  Avaliação
print("\nClassification Report (Logistic Regression com Undersampling):")
print(classification_report(y_test, y_pred))

# Matriz de Confusão
conf_matrix = confusion_matrix(y_test, y_pred, labels=y.unique())

plt.figure(figsize=(8,6))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Greens',
            xticklabels=y.unique(), yticklabels=y.unique())
plt.xlabel('Predito')
plt.ylabel('Real')
plt.title('Matriz de Confusão - Logistic Regression com Undersampling')
plt.show()
```
